```python
import numpy as np
import seaborn as sns
import pandas as pd
from sklearn import datasets
import matplotlib.pyplot as plt
```

```python
iris = datasets.load_iris()
```

```python
iris
```

```python
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names) 
#merubah data array menjadi data frame (tabel), sesuai urutan hrus pinter baca data mentah
```

```python
iris_df['species'] = iris.target_names[iris.target]
#menambah kolom ['species'] dari array iris.target
```

```python
iris_df.head()
```

```python
iris_df.shape
#shape ga pake kurung soalnya udah dataframe, klo pakai () tupel
#150 baris, 5 kolom
```

```python
iris_df.isnull().sum()
#aggregasi biasanya pakai ()
```

```python
categories = iris_df['species'].value_counts().reset_index()
categories
```

```python
# Membuat bar plot
plt.bar(categories['index'], categories['species'], color=['cyan', 'yellow', 'orange'])

# Menambahkan label dan judul
plt.xlabel('Species')
plt.ylabel('Count')
plt.title('Bar Plot Iris Count')

# Menampilkan plot
plt.show()
```

```python
iris_df
```

```python
print(iris_df.groupby(['species'])['sepal length (cm)'].mean())
print("-----------------------------")
print(iris_df.groupby(['species'])['sepal width (cm)'].mean())
print("-----------------------------")
print(iris_df.groupby(['species'])['petal length (cm)'].mean())
print("-----------------------------")
print(iris_df.groupby(['species'])['petal width (cm)'].mean())
```

```python
sns.boxplot(x='species', y='sepal length (cm)', data=iris_df)
plt.show()
#mengetahui range nilai satu kolom
```

```python
sns.boxplot(x='species', y='sepal width (cm)', data=iris_df)
plt.show()
```

```python
sns.boxplot(x='species', y='petal length (cm)', data=iris_df)
plt.show()
```

```python
sns.boxplot(x='species', y='petal width (cm)', data=iris_df)
plt.show()
```

```python
sns.scatterplot(x='sepal length (cm)', y='sepal width (cm)', hue='species', data=iris_df)
plt.show()
```

```python
sns.scatterplot(x='petal length (cm)', y='petal width (cm)', hue='species', data=iris_df)
plt.show()
```
